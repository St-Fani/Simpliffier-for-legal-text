```python
import json
import pandas as pd
import matplotlib.pyplot as plt
```

```python
with open("../data/all_v1.json") as f:
    tos_data = json.load(f)

tos_df = pd.DataFrame(tos_data.values())
tos_df = tos_df[["doc", "original_text", "reference_summary"]]
print(len(tos_df))
```

```python
print(tos_df.head())
tos_df.info()
```

```python

tos_df["summary_word_count"] = tos_df["reference_summary"].str.split().str.len()

tos_df.boxplot(column="summary_word_count")
plt.ylabel("words per summary")
plt.title("Length of ToS;DR reference summaries")
plt.savefig("../results/tos_summary_lengths.pdf")
plt.show()

print(tos_df["summary_word_count"].describe())
```
